# Write a function that adds up multiples of 9

-   the function should

    -   ask the user for input
    -   let the user know how to end the program
    -   check whether the input is a number that is a multiple of 9
    -   add multiples of 9 to a running total
    -   return the total

-   start with the most simple working version, then add complexities

-   this activity should take around 15 mins

```python
# Georgia Doing
# Practice Problem 1 for Practical Exam 2

def sum_nines():
    # takes user input, adding up multiples of 9
    # takes no parameters

    # initialize accumulator
    usr_sum = 0
    # set indicator variable to start loop
    go = True
    while go:
        # ask for a number or 'done'
        usr_in = input("Give me numbers and I'll count 9s. Or type 'done' to exit.")
        # check if loop should exit, robust to 'Done'
        if usr_in.lower() == "done":
            go = False
        # use mod operator to check if div by 9
        elif int(usr_in) % 9 == 0:
            usr_sum += int(usr_in)
        else:
            print("num not div by 9")
    # return rather than print answer
    return usr_sum

sum_nines()
```

# Change the function above to take an ascii value instead of a character

-   the function should:

    -   take a lists of words and an integer as parameters
    -   return a new list only containing words that start with the
        ascii value character

-   start with the most simple working version, then add complexities

-   this activity should take around 5 mins

```python
# Georgia Doing
# Practice Problem 3 for Practical Exam 2

def ascii_lists(alist, val):
    # filter lists of words to be alliterative
    # param alist: a list of words/strings
    # param value: an ascii value for selection
    # returns list

    new_list = []
    # check val is valid ascii
    if val < 0 or val > 255:
        print("val outside ascii range")
        return new_list
    # if val is valid, check against words
    for word in alist:
        if not type(word) == type("anystring"):
            print("skipping item", word)
        else:
            # convert from ascii value to character
            if word[0] == chr(val):
                new_list.append(word)
    return new_list

print(ascii_lists(["ginny","georgia","dog","goose",5],ord("g")))
```

```python
def ascii_lists(word_list, ascii_value):
    """take a word list (parameter 1) and an integer (parameter 2) and
        return a new list only containing words starting with the ascii value character"""
    # if the word list is empty, return None
    if word_list == []:
        return None
    # make a copy of the word list
    words = word_list.copy()
    # otherwise, create a new empty list
    new_list = []
    character = chr(ascii_value)
    # traverse through each element in the word list
    for element in words:
        # if the element is a list, delete the element (it is not a string)
        if element is list:
            del element
        # otherwise:
        else:
            # convert the element to a string (it could be int or bool)
            element = str(element)
            # if the first character in the element is the same as the character (parameter 2)
            if element[0] == character:
                # append the element to the new list
                new_list.append(element)

    # if there are no words in the word list that start with the character, return None
    if new_list == []:
        return None

    # return the list of words that start with the ascii value character
    return new_list


### BEGIN TESTING

words = ['apple', 'potato', 'arugula', 'armenia', [1, 2, 3], ['apple', 'potato'], True, False]
ascii_a = ord('a')
ascii_p = ord('p')
print(ascii_lists([], ascii_a))
print(ascii_lists(words, ascii_a))
print(ascii_lists(words, ascii_p))
print(ascii_lists(words, 60))
```
